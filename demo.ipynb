# FoodieAgent demo
Shows the multi-step trace (tool calls, provider used) for 3 goals.

```python
from agent.core import FoodAgent
def show(reply, trace):
    for s in trace: print(f"[{s['provider']}] step {s['step']} -> {s['tool'] or 'FINAL'} {s.get('args','')}\n    {s.get('result','')[:200]}")
    print("\nAGENT:", reply, "\n" + "-"*60)
LAT, LON = 23.0225, 72.5714   # change to your coordinates
```

## Goal 1: mood -> follow-up -> search

```python
a = FoodAgent(); a.set_location(LAT, LON)
show(*a.chat("I'm in a sweet mood"))
show(*a.chat("pure veg, within 3 km, takeaway"))
```

## Goal 2: memory used later (no re-asking diet)

```python
show(*a.chat("Actually something spicy now, what's closest?"))
print(a.state['profile'])
```

## Goal 3: failover (bad Groq key -> Gemini answers)

```python
import os; os.environ['GROQ_API_KEY']='invalid'
b = FoodAgent(); b.set_location(LAT, LON)
show(*b.chat("Craving cheesy food, vegetarian, 2 km, dine-in"))
```
